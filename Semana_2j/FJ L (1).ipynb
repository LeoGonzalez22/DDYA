## Enunciado
Una tienda necesita organizar los precios de sus productos. La cantidad de productos puede variar, por lo que el usuario debera indicar cuantos precios desea registrar.

El programa debe solicitar:

1.La cantidad de productos, representada por n.
1.El precio de cada uno de los productos.
3.El tipo de ordenamiento que se desea mostrar.

RQFO2 Leer el valor de cada producto
RQFO3 Leer la opcion de ordenamiento (OA, OD)
RQFO4: Ordenar los precios de los productos de forma ascendente o descendente segun la opcion seleccionada.
RQFO5: Mostrar los precios ordenados.

RQNFO1 El sistema debe permitir ingresar una cantidad n de productos variables
RQNFO2 El sistema no debe alterar los valores de los productos 
RQNFO3 El sistema debe manejar el criterio de ordenamiento elegido por el usuario
