Incluye:
La complejidad de este algoritmo es de O(n) ya que presenta un unico ciclo while el cual se va a repetir n-1 veces en todo el proceso, Las operaciones realizadas dentro del ciclo tienen una complejidad constante O(1). Por lo tanto, la complejidad total del algoritmo es O(n).

## 6) Código funcional **[15 pts]**



```python
while True:
    try:
        n=int(input("Ingrese el mes de visualizacion: "))
        año=[0,1]
        if n>=2:
            m=2
            while n>=m:
                sub=((año[m-1])+(año[m-2]))
                
                m+=1
                año.append(sub)#Meoizacion
        print(f"En el mes {n} habrian {sub} suscriptores")
            
    except ValueError:
        print("DATO NO VALIDO :(")
```
