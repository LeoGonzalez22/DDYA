### 5) Análisis de complejidad **[15 pts]**

- **f[x]=x**
- **g(n)=1**


Para buscar, insertar, eliminar y actualizar, para calcular la posición es una sola operación que no depende de n. Lo único que cambia es el recorrido de la lista de esa posición. Si hay pocos choques revisa uno o dos productos y es O(1). Si todos chocan, revisa los n y es O(n). En mostrar siempre se recorren las m posiciones y los n productos una vez, por eso no cambia entre mejor y peor caso.



## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def buscar(tabla, cod):
    pos = cod % len(tabla)
    for p in tabla[pos]:
        if p[0] == cod:
            return p
    return None

def insertar(tabla, p):
    pos = p[0] % len(tabla)
    if buscar(tabla, p[0]) != None:
        print("Ese codigo ya existe")
    else:
        tabla[pos].append(p)

def eliminar(tabla, cod):
    pos = cod % len(tabla)
    i = 0
    while i < len(tabla[pos]):
        if tabla[pos][i][0] == cod:
            tabla[pos].pop(i)
            return True
        i = i + 1
    return False

def mostrar(tabla):
    print("TABLA HASH")
    for i in range(len(tabla)):
        print(i, "->", end=" ")
        if len(tabla[i]) == 0:
            print("Vacio")
        else:
            for p in tabla[i]:
                print(p[0], "-", p[1], "- $", p[2], "-", p[3], "unidades", end=" | ")
            print()

m = int(input("Tamano de la tabla: "))
while m <= 0:
    print("El tamano debe ser mayor que cero")
    m = int(input("Tamano de la tabla: "))
tabla = []
for i in range(m):
    tabla.append([])

n = int(input("Cuantos productos desea registrar: "))
for i in range(n):
    print("Producto", i + 1)
    cod = int(input("Codigo: "))
    nom = input("Nombre: ")
    precio = int(input("Precio: "))
    cant = int(input("Cantidad: "))
    insertar(tabla, [cod, nom, precio, cant])

mostrar(tabla)

cod = int(input("Codigo que desea buscar: "))
p = buscar(tabla, cod)
if p == None:
    print("El producto no esta registrado")
else:
    print("Producto encontrado:", p[0], p[1], "Precio:", p[2], "Cantidad:", p[3])

cod = int(input("Codigo para actualizar cantidad: "))
p = buscar(tabla, cod)
if p == None:
    print("El producto no esta registrado")
else:
    nueva = int(input("Nueva cantidad: "))
    p[3] = nueva
    print("Inventario actualizado:", p[1], "-", p[3], "unidades")

cod = int(input("Codigo que desea eliminar: "))
if eliminar(tabla, cod):
    print("Producto eliminado")
else:
    print("No se encontro el producto")

mostrar(tabla)
```
